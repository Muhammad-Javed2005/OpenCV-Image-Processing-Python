```python
import cv2
```

```python
image = input("Enter image path: ")
print("1. Display Image"
      "\n2. Convert to Grayscale"
      "\n3. Get Image Dimensions"
      "\n4. Save Image"
      "\n5. Display Image")
choice = int(input("Enter your choice : " ))

```

```python
if choice == 1:
    img = cv2.imread(image)
    if img is not None:
        cv2.imshow("Image", img)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Image not found, cannot display image")

elif choice == 2:
    img = cv2.imread(image)
    if img is not None:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        cv2.imshow("Grayscale Image", gray)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Image not found, cannot convert to grayscale")

elif choice == 3:
    img = cv2.imread(image)
    if img is not None:
        height, width, channels = img.shape
        print(f"Image Dimensions: Width={width}, Height={height}, Channels={channels}")
    else:
        print("Image not found, cannot retrieve dimensions")

elif choice == 4:
    img = cv2.imread(image)
    if img is not None:
        save_path = input("Enter path to save the image: ")
        cv2.imwrite(save_path, img)
        print(f"Image saved to {save_path}")
    else:
        print("Image not found, cannot save image")

elif choice == 5:
    img = cv2.imread(image)
    if img is not None:
        cv2.imshow("Image", img)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Image not found, cannot display image")

else:
    print("Invalid choice")

```
